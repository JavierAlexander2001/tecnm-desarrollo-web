## 🧩 El reto

Define el contrato de un libro con un DTO de entrada y otro de salida. El servidor decide qué es válido antes de guardar.

**Requisitos del contrato**

- El modelo de entrada exige titulo (str), anio (int), precio (float) y disponible (bool).
- La coerción de "1967" a 1967 debe funcionar; "mil novecientos" debe fallar con 422.
- Los campos desconocidos se rechazan con 422 (extra="forbid").
- La respuesta nunca inventa campos: devuelve el id y los datos del libro.

> **Pista conceptual:** En Pydantic v2, `model_config = ConfigDict(extra="forbid")` convierte el DTO en un contrato cerrado.

````python
%%writefile app.py
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="API de libros")

class Libro(BaseModel):
    titulo: str
    anio: int
    precio: float
    disponible: bool

libros = []

@app.post("/api/libros", status_code=201)
def crear_libro(libro: Libro):
    nuevo_libro = {
        "id": len(libros) + 1,
        "titulo": libro.titulo,
        "anio": libro.anio,
        "precio": libro.precio,
        "disponible": libro.disponible
    }
    libros.append(nuevo_libro)
    return nuevo_libro

@app.get("/api/libros")
def listar_libros():
    return {"libros": libros}
````

````python
%%writefile app.py
from fastapi import FastAPI, status
from pydantic import BaseModel, ConfigDict

app = FastAPI(title="API de libros")


# DTO de entrada
class LibroEntrada(BaseModel):
    model_config = ConfigDict(extra="forbid")

    titulo: str
    anio: int
    precio: float
    disponible: bool


# DTO de salida
class LibroSalida(BaseModel):
    id: int
    titulo: str
    anio: int
    precio: float
    disponible: bool


libros: list[LibroSalida] = []


# Registrar un libro
@app.post(
    "/api/libros",
    response_model=LibroSalida,
    status_code=status.HTTP_201_CREATED
)
def crear_libro(libro: LibroEntrada) -> LibroSalida:
    nuevo_libro = LibroSalida(
        id=len(libros) + 1,
        titulo=libro.titulo,
        anio=libro.anio,
        precio=libro.precio,
        disponible=libro.disponible
    )

    libros.append(nuevo_libro)
    return nuevo_libro


# Consultar todos los libros
@app.get(
    "/api/libros",
    response_model=dict[str, list[LibroSalida]]
)
def listar_libros() -> dict[str, list[LibroSalida]]:
    return {"libros": libros}
````
