````python
%%writefile app.py

import time
from fastapi import FastAPI, Request

app = FastAPI()

# Middleware para medir el tiempo
@app.middleware("http")
async def middleware(request: Request, call_next):

    inicio = time.time()

    response = await call_next(request)

    tiempo = time.time() - inicio

    response.headers["X-Process-Time"] = str(tiempo)

    # Agregar identificador de petición
    request_id = request.headers.get("X-Request-ID")

    if request_id:
        response.headers["X-Request-ID"] = request_id

    return response


# Ruta para mostrar los libros
@app.get("/api/libros")
async def libros():
    return {"libros": []}

````

````python
%%writefile app.py
import time
import uuid

from fastapi import FastAPI, Request

app = FastAPI(title="Middleware Web")

# Middleware para medir el tiempo y controlar las peticiones
@app.middleware("http")
async def middleware(request: Request, call_next):

    inicio = time.perf_counter()

    # Obtener el identificador del cliente o generar uno nuevo
    request_id = request.headers.get("X-Request-ID")

    if not request_id:
        request_id = str(uuid.uuid4())

    # Continuar con la petición
    response = await call_next(request)

    # Calcular el tiempo de respuesta
    tiempo = time.perf_counter() - inicio

    # Agregar encabezados a todas las respuestas
    response.headers["X-Process-Time"] = f"{tiempo:.4f}"
    response.headers["X-Request-ID"] = request_id

    return response


# Endpoint para consultar los libros
@app.get("/api/libros")
async def obtener_libros():
    return {"libros": []}
````
