### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app_inicial.py
from fastapi import FastAPI

app = FastAPI()

libros = [
    {"id": 1, "titulo": "Rayuela", "autor": "Julio Cortázar"},
    {"id": 2, "titulo": "El principito", "autor": "Antoine de Saint-Exupéry"}
]

@app.get("/api/libros")
def listar_libros():
    return libros
````

````python
%%writefile app.py
from fastapi import FastAPI, status
from pydantic import BaseModel

app = FastAPI(
    title="Biblioteca API",
    version="1.0.0",
    description="API de libros del curso DAB-2201"
)

class Libro(BaseModel):
    id: int
    titulo: str
    autor: str

libros = []

@app.get(
    "/api/libros",
    response_model=dict[str, list[Libro]],
    summary="Listar libros",
    status_code=status.HTTP_200_OK
)
def listar_libros():
    return {"libros": libros}

@app.post(
    "/api/libros",
    response_model=Libro,
    summary="Crear un libro",
    status_code=status.HTTP_201_CREATED
)
def crear_libro(libro: Libro):
    libros.append(libro)
    return libro
````
