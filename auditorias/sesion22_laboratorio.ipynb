### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py

from fastapi import FastAPI

app = FastAPI(title="Mi biblioteca")

autores = []
libros = []

siguiente_autor = 1
siguiente_libro = 1


@app.post("/api/autores", status_code=201)
def crear_autor(datos: dict):
    global siguiente_autor

    autor = {
        "id": siguiente_autor,
        "nombre": datos["nombre"]
    }

    autores.append(autor)
    siguiente_autor += 1

    return autor


@app.post("/api/autores/{autor_id}/libros", status_code=201)
def crear_libro(autor_id: int, datos: dict):
    global siguiente_libro

    libro = {
        "id": siguiente_libro,
        "titulo": datos["titulo"],
        "anio": datos["anio"],
        "autor_id": autor_id
    }

    libros.append(libro)
    siguiente_libro += 1

    return libro


@app.get("/api/autores/{autor_id}")
def consultar_autor(autor_id: int):
    for autor in autores:
        if autor["id"] == autor_id:
            libros_autor = [
                {"id": l["id"], "titulo": l["titulo"]}
                for l in libros
                if l["autor_id"] == autor_id
            ]

            return {
                "id": autor["id"],
                "nombre": autor["nombre"],
                "libros": libros_autor
            }

    return {"error": "Autor no encontrado"}


@app.get("/api/libros")
def listar_libros():
    return {"libros": libros}

````

````python

%%writefile app.py
from fastapi import FastAPI, Depends, HTTPException
from pydantic import BaseModel
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey, event
from sqlalchemy.orm import declarative_base, relationship, sessionmaker, Session

# Configuracion de la base de datos
DATABASE_URL = "sqlite:///./biblioteca.db"

engine = create_engine(
    DATABASE_URL,
    connect_args={"check_same_thread": False}
)

# Activar llaves foraneas en SQLite
@event.listens_for(engine, "connect")
def activar_llaves_foreign_key(dbapi_connection, connection_record):
    cursor = dbapi_connection.cursor()
    cursor.execute("PRAGMA foreign_keys=ON")
    cursor.close()


SessionLocal = sessionmaker(
    autocommit=False,
    autoflush=False,
    bind=engine
)

Base = declarative_base()

app = FastAPI(title="Biblioteca con relaciones ORM")


# MODELOS ORM

class Autor(Base):
    __tablename__ = "autores"

    id = Column(Integer, primary_key=True, index=True)
    nombre = Column(String, nullable=False)

    libros = relationship(
        "Libro",
        back_populates="autor",
        cascade="all, delete-orphan"
    )


class Libro(Base):
    __tablename__ = "libros"

    id = Column(Integer, primary_key=True, index=True)
    titulo = Column(String, nullable=False)
    anio = Column(Integer, nullable=False)

    autor_id = Column(Integer, ForeignKey("autores.id"), nullable=False)

    autor = relationship("Autor", back_populates="libros")


# CREAR TABLAS AUTOMATICAMENTE
Base.metadata.create_all(bind=engine)


# MODELOS DE ENTRADA

class AutorEntrada(BaseModel):
    nombre: str


class LibroEntrada(BaseModel):
    titulo: str
    anio: int


# CONEXION A LA BASE DE DATOS

def obtener_db():
    db = SessionLocal()
    try:
        yield db
    finally:
        db.close()


# CREAR AUTOR

@app.post("/api/autores", status_code=201)
def crear_autor(datos: AutorEntrada, db: Session = Depends(obtener_db)):

    autor = Autor(nombre=datos.nombre)

    db.add(autor)
    db.commit()
    db.refresh(autor)

    return {
        "id": autor.id,
        "nombre": autor.nombre
    }


# CREAR LIBRO ASOCIADO A UN AUTOR

@app.post("/api/autores/{autor_id}/libros", status_code=201)
def crear_libro(
    autor_id: int,
    datos: LibroEntrada,
    db: Session = Depends(obtener_db)
):

    autor = db.query(Autor).filter(Autor.id == autor_id).first()

    if autor is None:
        raise HTTPException(status_code=404, detail="Autor no encontrado")

    libro = Libro(
        titulo=datos.titulo,
        anio=datos.anio,
        autor=autor
    )

    db.add(libro)
    db.commit()
    db.refresh(libro)

    return {
        "id": libro.id,
        "titulo": libro.titulo,
        "anio": libro.anio,
        "autor_id": libro.autor_id
    }


# CONSULTAR AUTOR Y SUS LIBROS

@app.get("/api/autores/{autor_id}")
def consultar_autor(autor_id: int, db: Session = Depends(obtener_db)):

    autor = db.query(Autor).filter(Autor.id == autor_id).first()

    if autor is None:
        raise HTTPException(status_code=404, detail="Autor no encontrado")

    return {
        "id": autor.id,
        "nombre": autor.nombre,
        "libros": [
            {
                "id": libro.id,
                "titulo": libro.titulo
            }
            for libro in autor.libros
        ]
    }


# LISTAR TODOS LOS LIBROS

@app.get("/api/libros")
def listar_libros(db: Session = Depends(obtener_db)):

    libros = db.query(Libro).all()

    return {
        "libros": [
            {
                "id": libro.id,
                "titulo": libro.titulo
            }
            for libro in libros
        ]
    }
````
